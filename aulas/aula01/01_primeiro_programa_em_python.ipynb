# **Primeiro Programa em Python**

Como uma tradição, é comum que programadores iniciem um programa ou comecem a aprender uma nova linguagem com um código de "Olá mundo" ou "Hello world", em inglês.

Em Python, esse programa é extremamente simples, comportando apenas uma única linha e usando a função `print()`.

Dentro da função `print()`, inserimos o parâmetro de texto, escrito entre aspas simples `''`, aspas duplas `""`, três pares de aspas simples `''''''` ou ainda três pares de aspas duplas `""""""`.

Em resumo, ao utilizar três pares de aspas simples ou duplas, é possível imprimir texto em diferentes linhas, mas isso será explicado melhor no tópico [`Entrada e Saída de Dados`](entrada_e_saida_de_dados.ipynb).

Vejamos o exemplo no código abaixo:

```python
# Ex. 01.a
print("Hello world!")
```

```python
# Ex. 01.b
print('Hello world!')
```

```python
# Ex. 01.c
print("""Hello world!""")
```

```python
# Ex. 01.d
print('''Hello world!''')
```
