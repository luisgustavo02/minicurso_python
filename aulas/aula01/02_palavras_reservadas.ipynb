# **Palavras Reservadas:**

Em Python, também temos uma lista de palavras reservadas, exclusivas para uso de funções, declaração de variáveis e outros artifícios. São eles:

- **False**
- **None**
- **True**
- **and**
- **as**
- **assert**
- **async**
- **await**
- **break**
- **class**
- **continue**
- **def**
- **del**
- **elif**
- **else**
- **except**
- **finally**
- **for**
- **from**
- **global**
- **if**
- **import**
- **in**
- **is**
- **lambda**
- **nonlocal**
- **not**
- **or**
- **pass**
- **raise**
- **return**
- **try**
- **while**
- **with**
- **yield**

Também não é de muito bom tom e não condiz com as [**boas práticas**](../aula00/recomendacoes_e_dicas.md) utilizar de outras palavras que são comumente utilizadas em Python:

- **NoneType**
- **dict**
- **float**
- **int**
- **list**
- **str**
- **tuple**
- **type**
